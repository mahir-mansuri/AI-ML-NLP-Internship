# Task 25: Manual Final CSV Export

## Objective
To export the cleaned heart disease dataset using a custom CSV-writing
function. The function will handle proper CSV formatting, UTF-8 encoding,
and file size optimization without using built-in DataFrame export methods.

```python
import pandas as pd
import csv
import os

df = pd.read_csv("../../datasets/heart.csv")

print("Dataset shape:", df.shape)
print(df.head())
print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Remove duplicate rows
final_df = df.drop_duplicates().reset_index(drop=True)

print("Original rows:", len(df))
print("Rows after removing duplicates:", len(final_df))
print("Columns:", list(final_df.columns))
```

## Custom CSV Export Function

The following function writes column names and data rows directly
using Python's csv module. It uses UTF-8 encoding and minimal quoting
to keep the output correctly formatted and avoid unnecessary file size.

```python
def manual_csv_export(dataframe, output_path):
    with open(
        output_path,
        mode="w",
        newline="",
        encoding="utf-8"
    ) as file:

        writer = csv.writer(
            file,
            quoting=csv.QUOTE_MINIMAL,
            lineterminator="\n"
        )

        # Write column names
        writer.writerow(dataframe.columns.tolist())

        # Write each data row
        for row in dataframe.itertuples(index=False, name=None):
            writer.writerow(row)

    print("CSV exported successfully:", output_path)


output_path = "../../datasets/heart_final_cleaned.csv"

manual_csv_export(final_df, output_path)
```

```python
file_size = os.path.getsize(output_path)

print("File size:", file_size, "bytes")

# Read the exported file back for validation
verified_df = pd.read_csv(output_path)

print("Exported dataset shape:", verified_df.shape)
print(verified_df.head())

if verified_df.shape == final_df.shape:
    print("Verification passed: row and column counts match.")
else:
    print("Verification failed: dataset dimensions differ.")
```

## Conclusion

The cleaned heart disease dataset was exported to CSV using a custom
function and Python's csv module instead of the DataFrame export method.
UTF-8 encoding, minimal CSV quoting, and newline handling were configured
to maintain compatibility and avoid unnecessary formatting overhead.
The exported file was then reloaded to verify its dimensions.
