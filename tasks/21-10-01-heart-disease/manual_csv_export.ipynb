```python
import pandas as pd
import csv

df = pd.read_csv("../../datasets/heart.csv")

print("Dataset shape:", df.shape)
print(df.head())
```

```python
processed_df = df.copy()

if "age" in processed_df.columns:
    values = [
        float(value)
        for value in processed_df["age"]
        if pd.notna(value)
    ]

    if values:
        age_mean = sum(values) / len(values)

        processed_df["age"] = processed_df["age"].apply(
            lambda value: age_mean if pd.isna(value) else value
        )

print("Processing completed.")
print("Rows:", len(processed_df))
```

```python
output_path = "../../datasets/heart_processed_manual.csv"

with open(output_path, "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(processed_df.columns.tolist())

    for row in processed_df.itertuples(index=False, name=None):
        writer.writerow([
            "" if pd.isna(value) else value
            for value in row
        ])

print("CSV saved:", output_path)
```

```python
with open(output_path, "r", newline="", encoding="utf-8") as file:
    exported_data = list(csv.reader(file))

print("Exported rows (including header):", len(exported_data))
print("Original rows:", len(df))
print("Exported columns:", len(exported_data[0]))
print("Header:", exported_data[0])
print("First exported record:", exported_data[1] if len(exported_data) > 1 else "No records")
```
