```python
# Cell 1: Load the dataset manually

file_path = "../../datasets/heart.csv"

with open(file_path, "r", encoding="utf-8-sig") as file:
    lines = file.read().splitlines()

headers = lines[0].split(",")

data = []

for line in lines[1:]:
    if line.strip():
        row = line.split(",")
        if len(row) == len(headers):
            data.append(row)

print("Dataset loaded successfully!")
print("Total rows:", len(data))
print("Total columns:", len(headers))
print("Column names:", headers)
```

```python
# Cell 2: Identify numerical columns manually

numeric_columns = []

for column_index, column_name in enumerate(headers):
    is_numeric = True
    found_value = False

    for row in data:
        value = row[column_index].strip()

        if value == "":
            continue

        found_value = True

        try:
            float(value)
        except ValueError:
            is_numeric = False
            break

    if is_numeric and found_value:
        numeric_columns.append(column_name)

print("Numerical columns:", numeric_columns)
```

```python
# Cell 3: Extract values from a selected column

def get_numeric_values(column_name):
    column_index = headers.index(column_name)
    values = []

    for row in data:
        value = row[column_index].strip()

        if value == "":
            continue

        try:
            values.append(float(value))
        except ValueError:
            continue

    return values


# Test with age column
values = get_numeric_values("age")

print("First 10 values:", values[:10])
print("Valid numeric values:", len(values))
```

```python
# Cell 4: Manual mean

def calculate_mean(values):
    total = 0
    count = 0

    for value in values:
        total += value
        count += 1

    if count == 0:
        return None

    return total / count


print("Mean:", calculate_mean(values))
```

```python
# Cell 5: Manual sorting and median

def manual_sort(values):
    result = []

    for value in values:
        position = 0

        while position < len(result) and result[position] <= value:
            position += 1

        result.insert(position, value)

    return result


def calculate_median(values):
    if len(values) == 0:
        return None

    ordered = manual_sort(values)
    n = len(ordered)
    middle = n // 2

    if n % 2 == 1:
        return ordered[middle]

    return (ordered[middle - 1] + ordered[middle]) / 2


print("Median:", calculate_median(values))
```

```python
# Cell 6: Manual mode

def calculate_mode(values):
    if len(values) == 0:
        return []

    frequency = {}

    for value in values:
        if value in frequency:
            frequency[value] += 1
        else:
            frequency[value] = 1

    highest_frequency = 0

    for value in frequency:
        if frequency[value] > highest_frequency:
            highest_frequency = frequency[value]

    modes = []

    for value in frequency:
        if frequency[value] == highest_frequency:
            modes.append(value)

    return manual_sort(modes)


print("Mode:", calculate_mode(values))
```

```python
# Cell 7: Manual minimum and maximum

def calculate_min(values):
    if len(values) == 0:
        return None

    smallest = values[0]

    for value in values:
        if value < smallest:
            smallest = value

    return smallest


def calculate_max(values):
    if len(values) == 0:
        return None

    largest = values[0]

    for value in values:
        if value > largest:
            largest = value

    return largest


print("Minimum:", calculate_min(values))
print("Maximum:", calculate_max(values))
```

```python
# Cell 8: Manual population standard deviation

def calculate_std(values):
    if len(values) == 0:
        return None

    mean = calculate_mean(values)
    squared_total = 0

    for value in values:
        difference = value - mean
        squared_total += difference * difference

    variance = squared_total / len(values)

    if variance == 0:
        return 0.0

    estimate = variance if variance >= 1 else 1.0

    for _ in range(100):
        estimate = (estimate + variance / estimate) / 2

    return estimate


print("Population standard deviation:", calculate_std(values))
```

```python
# Cell 9: Display all statistics

def calculate_all_statistics(values):
    return {
        "Count": len(values),
        "Mean": calculate_mean(values),
        "Median": calculate_median(values),
        "Mode": calculate_mode(values),
        "Minimum": calculate_min(values),
        "Maximum": calculate_max(values),
        "Population Std": calculate_std(values)
    }


age_values = get_numeric_values("age")

age_statistics = calculate_all_statistics(age_values)

for statistic, result in age_statistics.items():
    print(f"{statistic}: {result}")
```

```python
# Cell 10: Calculate statistics for every numerical column

all_results = {}

for column in numeric_columns:
    column_values = get_numeric_values(column)
    all_results[column] = calculate_all_statistics(column_values)

for column, results in all_results.items():
    print("\nColumn:", column)
    print("-" * 40)

    for statistic, result in results.items():
        print(f"{statistic}: {result}")
```
