```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../../datasets/heart.csv")

print("Dataset shape:", df.shape)
print(df.head())
```

```python
numerical_columns = [
    col for col in ["age", "trestbps", "chol", "thalach", "oldpeak"]
    if col in df.columns
]

print("Numerical columns:", numerical_columns)

print("\nSummary statistics:")
print(df[numerical_columns].describe())
```

```python
for col in numerical_columns:
    values = df[col].dropna()
    mean_value = values.mean()

    plt.figure(figsize=(7, 4))

    plt.hist(values, bins=20, edgecolor="black")
    plt.axvline(mean_value, linestyle="--",
                label=f"Mean = {mean_value:.2f}")

    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.annotate(
        f"Mean: {mean_value:.2f}",
        xy=(mean_value, 0.9),
        xycoords=("data", "axes fraction"),
        xytext=(5, -10),
        textcoords="offset points"
    )
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
plot_data = [
    df[col].dropna()
    for col in numerical_columns
]

plt.figure(figsize=(9, 5))

plt.boxplot(plot_data, tick_labels=numerical_columns)

plt.title("Comparison of Numerical Columns")
plt.xlabel("Numerical Features")
plt.ylabel("Values")
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
if "age" in df.columns and "chol" in df.columns:
    plot_df = df[["age", "chol"]].dropna()

    plt.figure(figsize=(8, 5))

    plt.scatter(plot_df["age"], plot_df["chol"], alpha=0.6)

    max_index = plot_df["chol"].idxmax()
    max_point = df.loc[max_index]

    plt.annotate(
        f"Highest chol: {max_point['chol']}",
        xy=(max_point["age"], max_point["chol"]),
        xytext=(10, 10),
        textcoords="offset points",
        arrowprops=dict(arrowstyle="->")
    )

    plt.title("Age vs Cholesterol")
    plt.xlabel("Age")
    plt.ylabel("Cholesterol (chol)")
    plt.tight_layout()
    plt.show()
```
