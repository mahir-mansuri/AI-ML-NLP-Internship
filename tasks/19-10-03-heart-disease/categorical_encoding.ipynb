```python
import pandas as pd

df = pd.read_csv("../../datasets/heart.csv")

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
print(df.head())
```

```python
categorical_columns = ["sex", "cp", "thal"]

for col in categorical_columns:
    if col in df.columns:
        print(f"\nColumn: {col}")
        print("Unique values:", df[col].unique())
        print("Number of categories:", df[col].nunique())
```

```python
def manual_encode(value, categories):
    if pd.isna(value):
        return -1

    for encoded_value, category in enumerate(categories):
        if value == category:
            return encoded_value

    return -1


encoding_maps = {
    "sex": [0, 1],
    "cp": [0, 1, 2, 3],
    "thal": [0, 1, 2, 3]
}

df_encoded = df.copy()

for col, categories in encoding_maps.items():
    if col in df_encoded.columns:
        df_encoded[col] = df_encoded[col].apply(
            lambda value: manual_encode(value, categories)
        )

print("Manual encoding completed!")
display(df_encoded.head())
```

```python
for col in categorical_columns:
    if col in df.columns:
        print(f"\n{col}")
        print("Before encoding:", sorted(df[col].dropna().unique().tolist(), key=str))
        print("After encoding:", sorted(df_encoded[col].unique().tolist(), key=str))
```

```python
test_categories = [0, 1, 2, 3]

test_values = [0, 1, 2, 3, 99, None]

for value in test_values:
    result = manual_encode(value, test_categories)
    print(f"Input: {value} -> Encoded: {result}")
```

```python
output_path = "../../datasets/heart_encoded.csv"

df_encoded.to_csv(output_path, index=False)

print("Encoded dataset saved:", output_path)
print("Original shape:", df.shape)
print("Encoded shape:", df_encoded.shape)
```

# Task 6: Manual Categorical Encoding

## Objective
To convert categorical values into numerical representations using a manually implemented encoding function without built-in encoding utilities.

## Methodology
1. Loaded the Heart Disease dataset using Pandas.
2. Identified the categorical columns and inspected their unique values.
3. Implemented a custom encoding function using explicit category comparisons.
4. Assigned numerical codes to known categories.
5. Handled unknown and missing values using a designated code of -1.
6. Verified the encoded values and saved the result as a separate CSV file.

## Unknown Value Handling
The custom function returns -1 when a value is missing or does not match any category in the defined mapping. This makes unexpected values identifiable, although -1 should be treated carefully during machine learning because it can be interpreted as an additional numerical value.

## Conclusion
Manual categorical encoding was implemented and tested without using built-in encoding functions. The encoded dataset was saved separately, preserving the original dataset.

The category mappings must be checked against the actual dataset values before using the encoded data for model training.
